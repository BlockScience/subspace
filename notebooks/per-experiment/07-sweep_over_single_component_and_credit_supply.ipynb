```python
df = pd.read_pickle('../../data/simulations/sweep_over_single_component_and_credit_supply-2023-12-29_21-27-45.pkl.gz').set_index('days_passed').bfill().fillna(0)

df.head()
```

```python
# Take numeric columns for normalizationn
df_numeric = df.select_dtypes(include=['number'])

# Create the normalized results
df_normalized = df_numeric / df_numeric.max()

# Add the label column back
df_normalized[['label', 'environmental_label']] = df[['label', 'environmental_label']]

# Drop unecessary columns
df_normalized = df_normalized.drop(['run', 'timestep', 'simulation', 'subset', 'timestep_in_days', 'block_time_in_seconds', 'sum_of_stocks', 'buffer_size'], axis=1).fillna(0)
df_normalized.shape
```

```python
df_normalized.hvplot.line(x='days_passed', groupby='label', by='environmental_label')
```
